
# Modelagem de máximos de precipitação na Zona da Mata mineira

### 70ª Reunião Brasileira de Estatística (RBras)

## Descrição

Este notebook reúne os procedimentos computacionais utilizados na análise de precipitações extremas na Zona da Mata mineira, considerando o período de 1961 a 2025.

Os dados de precipitação são provenientes da base Brazilian Daily Weather Gridded Data (BR-DWGD), versão 3.2.4. A área de estudo compreende 312 pixels com resolução espacial de 0,1° × 0,1°.

## Procedimentos estatísticos

- Cálculo dos máximos anuais de precipitação diária.
- Aplicação dos testes de Ljung–Box, Dickey–Fuller aumentado (ADF) e Kwiatkowski–Phillips–Schmidt–Shin (KPSS).
- Ajuste da Distribuição Generalizada de Valores Extremos (GEV), considerando modelos estacionário e não estacionários com tempo e El Niño.
- Seleção dos modelos pelo Critério de Informação de Akaike corrigido (AICc).
- Avaliação dos ajustes pelo teste de Kolmogorov–Smirnov (KS).
- Estimativa dos níveis de retorno e das probabilidades anuais de excedência.

## Resultados disponibilizados

O processamento gera tabelas com os resultados estatísticos e mapas referentes à área de estudo, aos modelos selecionados, à avaliação dos ajustes, aos níveis de retorno e às probabilidades de excedência de três magnitudes de precipitação diária.

As figuras são disponibilizadas nos formatos PNG e PDF, acompanhadas dos resultados numéricos em CSV.

```python
# ============================================================================
# 70ª RBras | REPRODUÇÃO DA ANÁLISE DA ZONA DA MATA MINEIRA (1961–2025)
# CÉLULA ÚNICA: fontes públicas -> máximos anuais -> testes -> GEV -> mapas
# NÃO MONTA NEM ACESSA O GOOGLE DRIVE PESSOAL DE NINGUÉM.
# ============================================================================

# 0. DEPENDÊNCIAS (a executar no Google Colab)
import sys, subprocess, shutil, os, re, warnings, zipfile
from pathlib import Path

subprocess.check_call([
    sys.executable, '-m', 'pip', 'install', '-q',
    'duckdb==1.5.5', 'geobr==2.1.1', 'geopandas', 'rioxarray',
    'xarray', 'netCDF4', 'dask', 'numpy', 'pandas', 'scipy', 'matplotlib',
    'shapely', 'beautifulsoup4', 'requests', 'gdown', 'lxml'
])

if not shutil.which('Rscript'):
    print('Instalando a linguagem R no ambiente temporário...')
    subprocess.check_call(['apt-get', 'update', '-qq'])
    subprocess.check_call(['apt-get', 'install', '-y', '-qq', 'r-base'])

import numpy as np
import pandas as pd
import xarray as xr
import rioxarray
import geopandas as gpd
import geobr
import requests
import gdown
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
from shapely.geometry import box, mapping
from matplotlib.patches import Patch, Rectangle, ConnectionPatch
from matplotlib.colors import Normalize, LogNorm
from matplotlib.cm import ScalarMappable
from scipy.stats import genextreme

warnings.filterwarnings('ignore', category=FutureWarning)

# 1. CONFIGURAÇÃO E FONTES PÚBLICAS
BASE = Path('/content/70rbras_reproducao')
DADOS = BASE / 'dados_entrada'
SAIDA = BASE / 'resultados'
for p in [DADOS, SAIDA]:
    p.mkdir(parents=True, exist_ok=True)

BR_SITE = 'https://sites.google.com/site/alexandrecandidoxavierufes/brazilian-daily-weather-gridded-data'
BR_PUBLIC_FOLDER = 'https://drive.google.com/drive/folders/11-qnvwojirAtaQxSE03N0_SUrbcsz44N'
NOAA_ONI_V5 = 'https://www.cpc.ncep.noaa.gov/products/analysis_monitoring/enso/oni/v5/'
ANO_INICIAL, ANO_FINAL = 1961, 2025
EVENTOS = [(110.6, '110,6 mm - Novo Aripuanã (AM)'),
           (128.4, '128,4 mm - Piranhas (AL)'),
           (138.6, '138,6 mm - Juiz de Fora (MG)')]
PERIODOS = ['19610101_19801231', '19810101_20001231', '20010101_20251231']
ARQUIVOS = [f'pr_{p}_BR-DWGD_UFES_UTEXAS_v_3.2.4.nc' for p in PERIODOS]

print('Fonte original BR-DWGD:', BR_SITE)
print('Fonte histórica ONI (NOAA):', NOAA_ONI_V5)
print('Destino dos resultados:', SAIDA)

# 2. DOWNLOAD DE DADOS PÚBLICOS DA BR-DWGD, SEM GOOGLE DRIVE PESSOAL
# Primeiro tenta o Google Drive público dos autores (arquivos ou ZIPs).
# Se falhar, tenta o Dropbox público listado na mesma página oficial.
# O Dropbox disponibiliza uma pasta; portanto a alternativa pode transferir
# um ZIP GRANDE, contendo mais variáveis do que as três de precipitação.
# Nenhuma dessas opções solicita autenticação ou upload do pesquisador.

from urllib.parse import urlsplit, urlunsplit, parse_qsl, urlencode

BR_DROPBOX = (
    'https://www.dropbox.com/scl/fo/t225fii1ir4o5ozga0o3u/'
    'ALZsq0F4zeCykN4GgoD8D6s?'
    'rlkey=93nkonzxn08c4wioztkjq0x52&dl=1'
)

def localizar_arquivos():
    return [DADOS / arq for arq in ARQUIVOS]

def faltantes():
    return [arq for arq in ARQUIVOS
            if not (DADOS / arq).is_file()
            or (DADOS / arq).stat().st_size == 0]

def extrair_nc_em_zip(arquivo_zip):
    encontrados = []
    with zipfile.ZipFile(arquivo_zip) as zf:
        for item in zf.infolist():
            nome = Path(item.filename).name
            if nome in faltantes():
                destino = DADOS / nome
                print('  Extraindo:', nome)
                with zf.open(item) as entrada, destino.open('wb') as saida:
                    shutil.copyfileobj(entrada, saida)
                encontrados.append(nome)

        # Algumas pastas distribuem arquivos ZIP menores dentro do ZIP maior.
        zip_internos = [x for x in zf.infolist() if x.filename.lower().endswith('.zip')]
        for item in zip_internos:
            if not faltantes():
                break
            nome = Path(item.filename).name
            if not any(chave in nome.lower() for chave in ['pr_', 'precip', '1961', '1981', '2001', 'brazil']):
                continue
            temporario = DADOS / ('_temp_' + nome)
            try:
                print('  Verificando ZIP interno:', nome)
                with zf.open(item) as entrada, temporario.open('wb') as saida:
                    shutil.copyfileobj(entrada, saida)
                if zipfile.is_zipfile(temporario):
                    with zipfile.ZipFile(temporario) as interno:
                        for x in interno.infolist():
                            n = Path(x.filename).name
                            if n in faltantes():
                                destino = DADOS / n
                                with interno.open(x) as entrada, destino.open('wb') as saida:
                                    shutil.copyfileobj(entrada, saida)
                                encontrados.append(n)
            finally:
                temporario.unlink(missing_ok=True)
    return encontrados

def baixar_google_publico():
    if not faltantes():
        return
    print('\n[1/2] Consultando a pasta pública da BR-DWGD no Google Drive...')
    try:
        catalogo = gdown.download_folder(
            url=BR_PUBLIC_FOLDER, output=str(DADOS),
            quiet=True, skip_download=True, remaining_ok=True)
        catalogo = catalogo or []
        print('Itens identificados na pasta pública:', len(catalogo))
        # Seleciona exatamente os NetCDF desejados ou ZIPs de precipitação.
        arquivos = [x for x in catalogo
                    if Path(x.path).name in faltantes()]
        zips = [x for x in catalogo
                if Path(x.path).suffix.lower() == '.zip'
                and any(k in Path(x.path).name.lower()
                        for k in ['pr_', 'precip', '1961', '1981', '2001'])]
        print('Arquivos diretamente identificados:', len(arquivos),
              '| ZIPs candidatos:', len(zips))
        for item in arquivos + zips:
            if not faltantes():
                break
            nome = Path(item.path).name
            destino = DADOS / nome
            print('  Baixando arquivo público:', nome)
            if not destino.is_file():
                r = gdown.download(id=item.id, output=str(destino),
                                   quiet=False, resume=True)
                if not r:
                    print('  O servidor não liberou:', nome)
                    continue
            if destino.suffix.lower() == '.zip':
                extrair_nc_em_zip(destino)
    except Exception as erro:
        print('Google Drive público indisponível:', str(erro)[:400])

def baixar_dropbox_publico():
    if not faltantes():
        return
    print('\n[2/2] Tentando a pasta oficial da BR-DWGD no Dropbox público...')
    print('Aviso: o Dropbox pode entregar toda a pasta como um ZIP grande.')
    destino_zip = DADOS / '_BR_DWGD_pasta_publica_dropbox.zip'
    try:
        if not zipfile.is_zipfile(destino_zip):
            # URL pública verificada na página dos autores; dl=1 força download.
            # Faz streaming para não ocupar RAM com o ZIP inteiro.
            with requests.get(BR_DROPBOX, stream=True, timeout=(30, 120),
                              headers={'User-Agent':'Mozilla/5.0'}) as r:
                r.raise_for_status()
                tamanho = int(r.headers.get('Content-Length', '0') or 0)
                disponivel = shutil.disk_usage(DADOS).free
                limite = min(30 * 1024**3, max(0, disponivel - 5 * 1024**3))
                if tamanho and tamanho > limite:
                    raise RuntimeError(
                        f'O arquivo da pasta pública tem {tamanho/1024**3:.1f} GB '
                        'e ultrapassa o limite de espaço desta sessão do Colab.')
                total = 0
                inicio = True
                with destino_zip.open('wb') as saida:
                    for bloco in r.iter_content(chunk_size=4 * 1024 * 1024):
                        if not bloco:
                            continue
                        if inicio:
                            inicio = False
                            if not bloco.startswith(b'PK'):
                                raise RuntimeError(
                                    'O Dropbox retornou uma página, não um arquivo ZIP. '
                                    'A transferência pública pode estar bloqueada.')
                        saida.write(bloco)
                        total += len(bloco)
                        if total > limite:
                            raise RuntimeError('Espaço insuficiente para baixar a pasta pública.')
                        if total // (512*1024**2) > (total-len(bloco)) // (512*1024**2):
                            print(f'  Recebidos {total/1024**3:.1f} GB...')
        if not zipfile.is_zipfile(destino_zip):
            raise RuntimeError('O Dropbox não enviou um ZIP válido.')
        achados = extrair_nc_em_zip(destino_zip)
        print('Arquivos de precipitação extraídos:', len(achados))
    except Exception as erro:
        print('Dropbox público indisponível:', str(erro)[:500])
    finally:
        # O arquivo compactado é temporário: mantemos apenas os NetCDF usados.
        # Não deletar se o arquivo for parcial é desnecessário, pois a próxima
        # tentativa fará o download completo novamente.
        destino_zip.unlink(missing_ok=True)

baixar_google_publico()
if faltantes():
    baixar_dropbox_publico()

if faltantes():
    raise RuntimeError(
        '\nNão foi possível obter automaticamente os arquivos públicos da BR-DWGD. '
        'Isso é um bloqueio das fontes de hospedagem, não uma falta de acesso ao seu Drive. '
        '\nArquivos ausentes: ' + ', '.join(faltantes()) +
        '\nFonte oficial: ' + BR_SITE +
        '\nÉ preciso disponibilizar links diretos estáveis para os três arquivos '
        'ou escolher uma forma alternativa de obter a grade. '
        'O notebook NÃO exige envio manual de arquivos e não continua com dados incompletos.'
    )

for arquivo in localizar_arquivos():
    print('✅', arquivo.name, f'({arquivo.stat().st_size/1024**2:.1f} MB)')
print('\nTrês arquivos BR-DWGD disponíveis; iniciando processamento.')

# 3. DELIMITAÇÃO GEOGRÁFICA DO IBGE (GEOBR) E MÁXIMOS ANUAIS
mg = geobr.read_state(code_state='MG', year=2020).to_crs('EPSG:4326')
meso = geobr.read_meso_region(code_meso='MG', year=2020)
zona = meso.loc[meso['name_meso'].eq('Zona da Mata')].to_crs('EPSG:4326')
if len(zona) != 1:
    raise RuntimeError('A mesorregião Zona da Mata não foi identificada unicamente.')

print('Recortando a Zona da Mata e calculando os máximos anuais...')
with xr.open_mfdataset([str(p) for p in localizar_arquivos()],
                       combine='by_coords', chunks={'time': 365}) as ds:
    ds = ds.sortby(['latitude', 'longitude'])
    ds.rio.write_crs('EPSG:4326', inplace=True)
    dz = ds.rio.clip(zona.geometry.apply(mapping), zona.crs, drop=True)
    anuais = dz['pr'].resample(time='YE').max(dim='time')
    df = anuais.to_dataframe().reset_index().rename(columns={'pr': 'Valor_Extremo'})

df['Ano'] = df['time'].dt.year
df = df.loc[df['Ano'].between(ANO_INICIAL, ANO_FINAL),
            ['latitude','longitude','Ano','Valor_Extremo']].dropna().copy()
df['Tempo'] = df['Ano'] - ANO_INICIAL + 1

# 4. ÍNDICE DE EL NIÑO: ONI HISTÓRICO V5 DA NOAA (ASO, SON, OND)
# Reproduz o procedimento original: média das três estações de primavera.
# A versão v5 da NOAA é mantida explicitamente, em vez de trocar silenciosamente
# por outra revisão do índice.
print('Obtendo o índice climático público da NOAA...')
resposta = requests.get(NOAA_ONI_V5, timeout=90,
                        headers={'User-Agent':'Mozilla/5.0'})
resposta.raise_for_status()
sopa = BeautifulSoup(resposta.text, 'html.parser')
linhas = []
for tr in sopa.find_all('tr'):
    col = [re.sub(r'\s+', ' ', td.get_text(' ', strip=True))
           for td in tr.find_all(['td','th'])]
    if len(col) >= 12 and re.fullmatch(r'\d{4}', col[0]):
        try:
            ano = int(col[0])
            linhas.append((ano, float(col[9]), float(col[10]), float(col[11])))
        except (ValueError, IndexError):
            pass
indice = pd.DataFrame(linhas, columns=['Ano','ASO','SON','OND'])
indice = indice.loc[indice['Ano'].between(ANO_INICIAL, ANO_FINAL)]
indice = indice.drop_duplicates('Ano').sort_values('Ano')
if len(indice) != 65:
    raise RuntimeError(f'Não foi possível recuperar os 65 anos da NOAA: {len(indice)} encontrados.')
indice['ENSO_Primavera'] = indice[['ASO','SON','OND']].mean(axis=1)
indice.to_csv(SAIDA/'indice_ENSO_primavera_NOAA_ONI_v5.csv', index=False)

valor_2025 = float(indice.loc[indice['Ano'].eq(2025), 'ENSO_Primavera'].iloc[0])
print('ENSO de referência em 2025:', valor_2025)
if abs(valor_2025 - (-0.5)) > 1e-8:
    raise RuntimeError('O valor da NOAA deixou de coincidir com a referência original (-0,5).'
                       ' Verifique a versão do índice antes de comparar os modelos.')

df_espacial_enso = df.merge(indice[['Ano','ENSO_Primavera']], on='Ano', how='inner')
df_espacial_enso = df_espacial_enso.sort_values(['latitude','longitude','Ano'])
# Identificador inteiro permanente nesta execução. Isso evita perder pixels
# ao cruzar resultados do R com a grade do Python por floats de lat/lon.
df_espacial_enso['pixel_id'] = (
    df_espacial_enso.groupby(['latitude','longitude'], sort=True).ngroup() + 1
)
n_pixels = df_espacial_enso['pixel_id'].nunique()
print(f'Base processada: {n_pixels} pixels e {df_espacial_enso.Ano.nunique()} anos.')
if n_pixels != 312:
    print('ATENÇÃO: o estudo original tinha 312 pixels. Verifique a versão da grade e o recorte.')

df_espacial_enso.to_csv(SAIDA/'maximos_anuais_1961_2025.csv', index=False)

# 5. EXECUTAR NO R OS MESMOS TESTES E AJUSTES DO NOTEBOOK ORIGINAL
# São preservados:
#  - Ljung-Box, ADF e KPSS (nível e tendência);
#  - GEV estacionária e GEV com localização variando com tempo ou ENSO;
#  - elegibilidade do modelo temporal mediante KPSS tendência p>0,05;
#  - seleção AICc com preferência por estacionário se delta <=2;
#  - transformação integral de probabilidade e teste KS;
#  - nível de retorno no último ano (2025).
# Não há teste de Pettitt nem modelos com ruptura (retirados por serem
# alheios à análise final). O teste runs também não afetava o ajuste.

codigo_R = r'''
repos <- 'https://cloud.r-project.org'
necessarios <- c('extRemes','tseries')
ausentes <- setdiff(necessarios, rownames(installed.packages()))
if (length(ausentes)) install.packages(ausentes, repos=repos, quiet=TRUE)
suppressPackageStartupMessages(library(extRemes))
suppressPackageStartupMessages(library(tseries))

args <- commandArgs(trailingOnly=TRUE)
entrada <- args[1]; saida <- args[2]
dat <- read.csv(entrada)
dat <- dat[order(dat$latitude, dat$longitude, dat$Ano), ]

calcula_aicc <- function(fit) {
    k <- length(fit$results$par)
    n <- fit$n
    nll <- fit$results$value
    (2*k + 2*nll) + 2*k*(k+1)/(n-k-1)
}

por_pixel <- split(dat, dat$pixel_id, drop=TRUE)

res <- lapply(por_pixel, function(d) {
  y <- d$Valor_Extremo
  cov_t <- d$Tempo
  cov_e <- d$ENSO_Primavera
  lat <- d$latitude[1]; lon <- d$longitude[1]

  lj <- tryCatch(Box.test(y, type='Ljung-Box')$p.value, error=function(e) NA_real_)
  adf <- tryCatch(suppressWarnings(adf.test(y)$p.value), error=function(e) NA_real_)
  kl <- tryCatch(suppressWarnings(kpss.test(y, null='Level')$p.value), error=function(e) NA_real_)
  kt <- tryCatch(suppressWarnings(kpss.test(y, null='Trend')$p.value), error=function(e) NA_real_)

  fit_est <- NULL; fit_tempo <- NULL; fit_enso <- NULL
  escolhido <- 'Falha_Ajuste'; pks <- NA_real_
  mu_ref <- NA_real_; sigma <- NA_real_; xi <- NA_real_
  rl <- rep(NA_real_, 5); ciinf <- NA_real_; cisup <- NA_real_

  if (!is.na(adf) && length(y) >= 30) {
    notas <- c()
    fit_est <- tryCatch(fevd(y, type='GEV'), error=function(e) NULL)
    if (!is.null(fit_est)) notas['Estacionario'] <- calcula_aicc(fit_est)

    if (!is.na(kt) && kt > 0.05) {
      fit_tempo <- tryCatch(
        fevd(y, data=data.frame(Tempo=cov_t), location.fun=~Tempo, type='GEV'),
        error=function(e) NULL)
      if (!is.null(fit_tempo)) notas['Tempo_Linear'] <- calcula_aicc(fit_tempo)
    }

    fit_enso <- tryCatch(
      fevd(y, data=data.frame(ENSO_Primavera=cov_e),
           location.fun=~ENSO_Primavera, type='GEV'), error=function(e) NULL)
    if (!is.null(fit_enso)) notas['ENSO'] <- calcula_aicc(fit_enso)

    notas <- notas[is.finite(notas)]
    if (length(notas)) {
      escolhido <- names(sort(notas))[1]
      if ('Estacionario' %in% names(notas) && escolhido != 'Estacionario' &&
          notas['Estacionario'] - min(notas) <= 2) escolhido <- 'Estacionario'
      fit <- switch(escolhido, Estacionario=fit_est,
                    Tempo_Linear=fit_tempo, ENSO=fit_enso)
      pars <- fit$results$par
      sigma <- as.numeric(pars['scale'])
      xi <- as.numeric(pars['shape'])
      if (escolhido == 'Estacionario') {
        mu_anos <- rep(as.numeric(pars[1]), length(y))
        mu_ref <- as.numeric(pars[1])
      } else if (escolhido == 'Tempo_Linear') {
        mu_anos <- as.numeric(pars[1]) + as.numeric(pars[2])*cov_t
        mu_ref <- as.numeric(pars[1]) + as.numeric(pars[2])*tail(cov_t,1)
      } else {
        mu_anos <- as.numeric(pars[1]) + as.numeric(pars[2])*cov_e
        mu_ref <- as.numeric(pars[1]) + as.numeric(pars[2])*tail(cov_e,1)
      }
      pit <- mapply(function(yi,mi) pevd(q=yi, loc=mi,
                   scale=sigma, shape=xi, type='GEV'), y, mu_anos)
      # Semente fixa apenas para tornar reproduzível a pequena perturbação do KS
      set.seed(70)
      pks <- tryCatch(suppressWarnings(
        ks.test(jitter(pit,amount=1e-7), 'punif')$p.value),
        error=function(e) NA_real_)

      nivel <- function(T) {
        p <- 1/T
        if (abs(xi) < 1e-6) return(mu_ref-sigma*log(-log(1-p)))
        mu_ref - (sigma/xi)*(1 - (-log(1-p))^(-xi))
      }
      rl <- sapply(c(5,10,20,50,100), nivel)
      tryCatch({
        if (escolhido == 'Estacionario') {
          interval <- ci(fit, type='return.level',return.period=100,method='normal')
        } else if (escolhido == 'Tempo_Linear') {
          qv <- make.qcov(fit,vals=list(mu1=tail(cov_t,1)))
          interval <- ci(fit,type='return.level',return.period=100,
                         method='normal',qcov=qv)
        } else {
          qv <- make.qcov(fit,vals=list(mu1=tail(cov_e,1)))
          interval <- ci(fit,type='return.level',return.period=100,
                         method='normal',qcov=qv)
        }
        ciinf <- as.numeric(interval[1]); cisup <- as.numeric(interval[3])
      }, error=function(e) {})
    }
  }
  data.frame(pixel_id=as.integer(d$pixel_id[1]), latitude=lat, longitude=lon,
             Modelo_SEM_Ruptura=escolhido,
             Ljung_Box_p=lj, ADF_p=adf, KPSS_Level_p=kl, KPSS_Trend_p=kt,
             KS_p=pks, mu_cenario=mu_ref, sigma_cenario=sigma, xi_cenario=xi,
             RL_5=rl[1], RL_10=rl[2], RL_20=rl[3], RL_50=rl[4], RL_100=rl[5],
             RL100_Inf=ciinf, RL100_Sup=cisup)
})
resultado <- do.call(rbind,res)
write.csv(resultado,saida,row.names=FALSE)
cat('Modelos ajustados:', nrow(resultado),'pixels\n')
cat('Distribuição:\n')
print(table(resultado$Modelo_SEM_Ruptura))
'''

arquivo_r = BASE/'ajuste_gev.R'
arquivo_r.write_text(codigo_R, encoding='utf-8')
print('\nExecutando testes e ajustes GEV em R...')
subprocess.run(['Rscript', str(arquivo_r),
                str(SAIDA/'maximos_anuais_1961_2025.csv'),
                str(SAIDA/'resultados_modelos_completos.csv')], check=True)

result = pd.read_csv(SAIDA/'resultados_modelos_completos.csv')
if (len(result) != n_pixels or result['pixel_id'].duplicated().any()
        or result['pixel_id'].isna().any()):
    raise RuntimeError(
        f'Resultados R incompletos ou duplicados: {len(result)} linhas; '
        f'{n_pixels} pixels esperados.'
    )
validos = result.loc[result['Modelo_SEM_Ruptura'] != 'Falha_Ajuste'].copy()

# 6. PROBABILIDADES EXATAS DOS CENÁRIOS: PARÂMETROS DO AJUSTE R
for mm, _ in EVENTOS:
    sufixo = str(mm).replace('.', '_')
    col = 'p_' + sufixo
    result[col] = np.nan
    ok = ((result['sigma_cenario'] > 0)
          & result[['mu_cenario','sigma_cenario','xi_cenario']].notna().all(axis=1))
    result.loc[ok, col] = 100 * genextreme.sf(
        mm,
        c=-result.loc[ok,'xi_cenario'].to_numpy(),
        loc=result.loc[ok,'mu_cenario'].to_numpy(),
        scale=result.loc[ok,'sigma_cenario'].to_numpy())
    # T (anos) = 100 / p (%) sob as condições de 2025; não é espera futura real.
    col_tr = 'TR_' + sufixo
    result[col_tr] = np.nan
    positivos = result[col].gt(0)
    result.loc[positivos,col_tr] = 100 / result.loc[positivos,col]
    result.loc[result[col].eq(0),col_tr] = np.inf

result.to_csv(SAIDA/'resultados_modelos_e_cenarios.csv',index=False)

# 7. QUADRÍCULAS RECORTADAS PELA MESORREGIÃO PARA TODOS OS MAPAS
# Associação robusta por pixel_id; a comparação com floats causava mapas pontilhados
centros = df_espacial_enso[['pixel_id','latitude','longitude']].drop_duplicates().copy()
lats = np.sort(centros.latitude.unique()); lons = np.sort(centros.longitude.unique())
dlat = float(np.min(np.diff(lats))); dlon = float(np.min(np.diff(lons)))
centros['geometry'] = [box(lon-dlon/2,lat-dlat/2,lon+dlon/2,lat+dlat/2)
                       for lat,lon in zip(centros.latitude,centros.longitude)]
grade = gpd.GeoDataFrame(centros,geometry='geometry',crs='EPSG:4326')
# Une dados espaciais e estatísticos pela chave inteira, NÃO pelos floats.
# A latitude/longitude válida é a coordenada original do raster.
grade = grade.merge(
    result.drop(columns=['latitude','longitude']),
    on='pixel_id', how='left', validate='one_to_one', indicator=True
)
sem_modelo = grade.loc[grade['_merge'].ne('both'), 'pixel_id'].tolist()
if sem_modelo:
    raise RuntimeError(
        f'Falha ao associar resultados em {len(sem_modelo)} pixels: '
        f'{sem_modelo[:15]}. Não serão gerados mapas incompletos.'
    )
grade = grade.drop(columns=['_merge'])
grade = gpd.clip(grade,zona)
print(f'✅ Todos os {len(result)} resultados foram associados à grade.')
print('Pixels com modelo válido:', grade['Modelo_SEM_Ruptura'].ne('Falha_Ajuste').sum())
print('Pixels sem probabilidade KS:', grade['KS_p'].isna().sum())
for cc in ['RL_10','RL_20','RL_100']:
    print(f'Pixels com {cc}:', grade[cc].notna().sum())

xmin,ymin,xmax,ymax = zona.total_bounds

plt.rcParams.update({'font.family':'DejaVu Sans','font.size':12})

def acabamento(ax):
    zona.boundary.plot(ax=ax,color='#253746',linewidth=1.1,zorder=10)
    ax.set_xlim(xmin-.05,xmax+.05); ax.set_ylim(ymin-.05,ymax+.05)
    ax.set_aspect(1/np.cos(np.deg2rad((ymin+ymax)/2)))
    ax.set_axis_off()

def fundo(ax):
    grade.plot(ax=ax,color='#D9DEE3',edgecolor='white',linewidth=.25)

def salva(fig,nome):
    for ext in ('png','pdf'):
        fig.savefig(SAIDA/f'{nome}.{ext}',dpi=350,bbox_inches='tight',
                    pad_inches=.08,facecolor='white')
    plt.show(); plt.close(fig)
    print('Figura salva:',nome)

# FIGURA 1 - MINAS GERAIS + AMPLIAÇÃO DA ZONA DA MATA
fig,(a1,a2) = plt.subplots(1,2,figsize=(12,6.2),
                          gridspec_kw={'width_ratios':[1,1.25]})
mg.plot(ax=a1,color='#E6E6E6',edgecolor='#666666',linewidth=.8)
zona.plot(ax=a1,color='#5DADE2',edgecolor='#1F4E79',linewidth=1)
a1.add_patch(Rectangle((xmin-.15,ymin-.15),xmax-xmin+.3,ymax-ymin+.3,
                       fill=False,edgecolor='#1F4E79',lw=1.4,ls='--'))
a1.set_title('Minas Gerais',fontsize=15,fontweight='bold')
a1.set_axis_off(); a1.set_aspect('equal')
grade.plot(ax=a2,color='#8BC7DD',edgecolor='white',linewidth=.35)
acabamento(a2)
a2.set_title('Zona da Mata - ampliação',fontsize=15,fontweight='bold')
fig.subplots_adjust(wspace=.07)
salva(fig,'fig_area_estudo')

# FIGURA 2 - MODELO SELECIONADO (PERCENTUAIS NA LEGENDA)
cores_modelos = {'Estacionario':('Estacionário','#246EAB'),
                 'Tempo_Linear':('GEV com tempo','#E8762C'),
                 'ENSO':('GEV com El Niño','#329B5C')}
fig,ax=plt.subplots(figsize=(9.4,10.2)); fundo(ax)
legenda=[]
for m,(nome,cor) in cores_modelos.items():
    sub=grade.loc[grade['Modelo_SEM_Ruptura'].eq(m)]
    if len(sub): sub.plot(ax=ax,color=cor,edgecolor='white',linewidth=.35)
    frac=100*result['Modelo_SEM_Ruptura'].eq(m).sum()/max(len(validos),1)
    legenda.append(Patch(facecolor=cor,label=f'{nome} ({frac:.1f}%)'))
acabamento(ax)
fig.legend(handles=legenda,loc='lower center',bbox_to_anchor=(.5,.02),
           ncol=1,frameon=False,fontsize=14)
fig.subplots_adjust(bottom=.14)
salva(fig,'fig_modelos_gev')

# FIGURA 3 - KS
fig,ax=plt.subplots(figsize=(9.4,10.2)); fundo(ax)
ks_ok=grade.loc[grade['KS_p'].gt(.05)]
ks_nao=grade.loc[grade['KS_p'].le(.05)]
if len(ks_ok): ks_ok.plot(ax=ax,color='#198C78',edgecolor='white',linewidth=.35)
if len(ks_nao): ks_nao.plot(ax=ax,color='#DF654B',edgecolor='white',linewidth=.35)
acabamento(ax)
ks_total=max(len(validos),1)
ks_leg=[Patch(facecolor='#198C78',label=f'Sem rejeição ({100*validos.KS_p.gt(.05).sum()/ks_total:.1f}%)'),
        Patch(facecolor='#DF654B',label=f'Rejeição ({100*validos.KS_p.le(.05).sum()/ks_total:.1f}%)'),
        Patch(facecolor='#D9DEE3',label=f'Sem resultado ({100*validos.KS_p.isna().sum()/ks_total:.1f}%)')]
fig.legend(handles=ks_leg,loc='lower center',bbox_to_anchor=(.5,.02),ncol=1,frameon=False,fontsize=14)
fig.subplots_adjust(bottom=.14)
salva(fig,'fig_teste_ks')

# FUNÇÃO COMPARTILHADA: FIGURAS 4 E 5 (TRÊS MAPAS POR FILEIRA)
def tres_mapas(mapas,cmap,norm,label,nome,ticks=None,labels=None):
    fig=plt.figure(figsize=(16,5.8),facecolor='white')
    gs=fig.add_gridspec(2,3,height_ratios=[1,.07],left=.025,right=.975,
                        top=.94,bottom=.12,wspace=.06,hspace=.15)
    for i,(col,titulo) in enumerate(mapas):
        ax=fig.add_subplot(gs[0,i]); fundo(ax)
        dados=grade.dropna(subset=[col])
        if len(dados):
            dados.plot(ax=ax,column=col,cmap=cmap,norm=norm,
                       edgecolor='white',linewidth=.25,legend=False)
        acabamento(ax)
        ax.set_title(titulo,fontsize=15,fontweight='bold',pad=12)
    cbax=fig.add_subplot(gs[1,:])
    sm=ScalarMappable(norm=norm,cmap=cmap); sm.set_array([])
    cb=fig.colorbar(sm,cax=cbax,orientation='horizontal',ticks=ticks)
    if labels is not None: cb.set_ticklabels(labels)
    cb.ax.minorticks_off()
    cb.set_label(label,fontsize=13); cb.ax.tick_params(labelsize=11)
    salva(fig,nome)

# FIGURA 4 - NÍVEIS DE RETORNO (10, 20 E 100 ANOS)
mapas4=[('RL_10','10% - 10 anos'),('RL_20','5% - 20 anos'),
        ('RL_100','1% - 100 anos')]
valores=np.concatenate([grade[c].dropna().to_numpy() for c,_ in mapas4])
tres_mapas(mapas4,plt.get_cmap('YlGnBu'),
           Normalize(vmin=float(valores.min()),vmax=float(valores.max())),
           'Precipitação máxima diária (mm)','fig_niveis_retorno_3mapas')

# FIGURA 5 - PERÍODO DE RETORNO EQUIVALENTE (ESCALA CONTÍNUA LOGARÍTMICA)
mapas5=[]
for mm,titulo in EVENTOS:
    tag=str(mm).replace('.','_')
    origem='TR_'+tag; visual='TR_visual_'+tag
    # Infinito numérico por p=0 é apenas um limite de representação.
    # Usa a classe mais alta (>= 1000) sem afirmar retorno infinito real.
    grade[visual]=grade[origem].clip(lower=10,upper=1000)
    mapas5.append((visual,titulo))
tres_mapas(mapas5,plt.get_cmap('YlOrRd'),
           LogNorm(vmin=10,vmax=1000,clip=True),
           'Período de retorno equivalente (anos)',
           'fig_retorno_equivalente_3mapas',
           ticks=[10,20,50,100,200,500,1000],
           labels=['10','20','50','100','200','500','≥ 1.000'])

# FIGURA 6 (EXTRA) - PROBABILIDADE ANUAL DE EXCEDÊNCIA (%) DOS TRÊS EVENTOS
# Esta figura é complementar às cinco figuras do pôster e sai no mesmo ZIP.
# Usa diretamente as probabilidades calculadas com os parâmetros GEV em R,
# já associadas aos pixels por pixel_id. Não é necessário carregar outro arquivo.
mapas6=[]
for mm,titulo in EVENTOS:
    col='p_'+str(mm).replace('.','_')
    # Uma probabilidade válida pode ser exatamente zero na precisão numérica.
    # Não a confundimos com dado ausente (NaN).
    vals=pd.to_numeric(grade[col],errors='coerce')
    n_invalidos=int((vals.notna() & ~vals.between(0,100)).sum())
    if n_invalidos:
        raise ValueError(f'{col}: {n_invalidos} probabilidades fora de [0, 100]%.')
    grade[col]=vals
    n_sem=int(vals.isna().sum())
    if n_sem:
        print(f'ATENÇÃO: {col} tem {n_sem} pixels sem probabilidade; esses ficam cinza.')
    mapas6.append((col,titulo))

probs_validas=np.concatenate([
    grade[col].dropna().to_numpy(dtype=float) for col,_ in mapas6
    if grade[col].notna().any()
]) if any(grade[col].notna().any() for col,_ in mapas6) else np.array([])
if probs_validas.size == 0:
    raise RuntimeError('Nenhuma probabilidade de excedência disponível para a figura extra.')
max_prob=float(probs_validas.max())
# Escala linear compartilhada por todos os mapas, sem classes artificiais.
# Exemplo: 5,252% de máximo -> legenda entre 0 e 6%.
max_escala=max(1.0,float(np.ceil(max_prob)))
ticks_prob=np.linspace(0,max_escala,7)
print(f'Figura extra: probabilidade máxima = {max_prob:.3f}%; escala 0 a {max_escala:g}%.')
tres_mapas(
    mapas6,
    plt.get_cmap('YlOrRd'),
    Normalize(vmin=0,vmax=max_escala),
    'Probabilidade anual de excedência (%)',
    'fig_probabilidade_evento_3mapas',
    ticks=ticks_prob,
    labels=[f'{x:g}' for x in ticks_prob]
)

# 8. RESULTADOS-CHAVE E CONFERÊNCIA COM O PÔSTER ORIGINAL
resumo=[]
def rel(s):
    print(s); resumo.append(s)
rel('70ª RBras - RESULTADOS DA EXECUÇÃO INDEPENDENTE')
rel('BR-DWGD v3.2.4; 1961–2025; recorte IBGE 2020; NOAA ONI v5')
rel(f'Pixels totais: {n_pixels}; com modelo ajustado: {len(validos)}')
rel(f'Referência GEV não estacionária: 2025 (Tempo=65; ENSO={valor_2025:.1f})')
rel(f'Ljung-Box p>0,05: {result.Ljung_Box_p.gt(.05).sum()}/{len(result)}')
rel(f'ADF p<0,05: {result.ADF_p.lt(.05).sum()}/{len(result)}')
rel(f'KPSS nível p>0,05: {result.KPSS_Level_p.gt(.05).sum()}/{len(result)}')
rel(f'KPSS tendência p>0,05: {result.KPSS_Trend_p.gt(.05).sum()}/{len(result)}')
for m in ['Estacionario','Tempo_Linear','ENSO']:
    n=validos.Modelo_SEM_Ruptura.eq(m).sum()
    rel(f'Modelo {m}: {n}/{len(validos)} ({100*n/max(len(validos),1):.1f}%)')
rel(f'KS sem rejeição p>0,05: {validos.KS_p.gt(.05).sum()}/{len(validos)}')
for t in [10,20,100]:
    v=validos[f'RL_{t}'].dropna()
    rel(f'Nível retorno {t} anos: {v.min():.1f} a {v.max():.1f} mm')
for mm,_ in EVENTOS:
    col='p_'+str(mm).replace('.','_')
    p=result[col].dropna()
    rel(f'Chuva {mm:.1f} mm: {len(p)} pixels; p anual {p.min():.3f}% a {p.max():.3f}%;'
        f' p<1% em {100*p.lt(1).mean():.1f}%; p<0,5% em {100*p.lt(.5).mean():.1f}%')

(SAIDA/'resumo_resultados.txt').write_text('\n'.join(resumo)+'\n',encoding='utf-8')

esperados={'Estacionario':190,'Tempo_Linear':120,'ENSO':2}
contagens=validos.Modelo_SEM_Ruptura.value_counts().to_dict()
if contagens != esperados:
    print('\nAVISO CIENTÍFICO: os modelos selecionados diferem dos resultados originais.')
    print('Original:',esperados,'| Reproduzido:',contagens)
    print('Não trate esta execução como reprodução exata sem comparar a tabela ENSO histórica'
          ' e conferir arquivos BR-DWGD/versões de pacotes.')
else:
    print('\nConferência: contagens de modelos coincidem com o pôster (190,120,2).')

# 9. DOCUMENTAÇÃO E ZIP, SEM ANEXAR BR-DWGD BRUTO
leia = f'''REPRODUÇÃO DO PÔSTER — 70ª RBras

FONTES ORIGINAIS
- Precipitação BR-DWGD v3.2.4, variável pr, NetCDF (1961–2025): {BR_SITE}
- Diretório público dos autores: {BR_PUBLIC_FOLDER}
- ENSO: Oceanic Niño Index (ONI) histórico ERSSTv5, NOAA CPC: {NOAA_ONI_V5}
- Geometria: IBGE 2020, mesorregião da Zona da Mata via pacote geobr.

COMO EXECUTAR
1. Abra este arquivo .ipynb em um Colab novo e execute sua célula única.
2. O código tenta baixar automaticamente os três arquivos de precipitação públicos.
3. Se as fontes públicas bloquearem o acesso automático, a execução será interrompida,
   indicando os arquivos ausentes e os endereços das fontes. Nenhum arquivo pessoal é solicitado.
4. Os resultados serão salvos em /content/70rbras_reproducao/resultados/.
   A execução NÃO requer acesso ao Drive pessoal da pesquisadora.

PROCEDIMENTO
Máximos diários anuais de 1961 a 2025 para cada pixel; testes Ljung-Box,
ADF, KPSS (nível/tendência); modelos GEV estacionário, com localização linear
no tempo ou no índice ENSO de primavera (média ASO, SON e OND); seleção por AICc,
preferência estacionária se delta <=2 e avaliação pelo KS. Períodos de retorno
condicionados aos valores de tempo e ENSO de 2025. Probabilidades dos eventos
foram calculadas diretamente a partir dos parâmetros GEV ajustados no R.
As Figuras 5 e 6 mostram, respectivamente, os retornos equivalentes e as
probabilidades anuais de excedência para três magnitudes observadas no Brasil.

INTERPRETAÇÃO
T = 1/p é período de retorno EQUIVALENTE, condicionado às covariáveis de 2025,
não previsão do intervalo real até uma futura chuva. Valor p=0 em precisão
numérica é exibido na classe visual mais alta, não prova de impossibilidade.
O KS aplicado sobre valores transformados pelo próprio ajuste não incorpora
uma correção de distribuição nula para parâmetros estimados dos mesmos dados.

VERIFICAÇÃO
Compare os resultados desta execução aos valores do pôster. A planilha original
El niño.xlsx não estava anexada ao notebook de origem; foi substituída pelo
índice ONI v5 público da NOAA, calculado com o mesmo critério ASO/SON/OND.
Apesar de produzir ENSO= -0,5 em 2025, a identidade de toda a série histórica
precisa ser confirmada antes de declarar reprodução numérica exata.

CITAÇÃO DOS DADOS
Xavier, A. C., Scanlon, B. R., King, C. W. & Alves, A. I. (2022).
New improved Brazilian daily weather gridded data (1961–2020).
International Journal of Climatology, 42(16), 8390–8404.
https://doi.org/10.1002/joc.7731
'''
(SAIDA/'LEIA-ME.txt').write_text(leia,encoding='utf-8')
zipfinal=BASE/'70_RBras_resultados_sem_dados_brutos.zip'
with zipfile.ZipFile(zipfinal,'w',compression=zipfile.ZIP_DEFLATED) as z:
    for arquivo in SAIDA.iterdir():
        if arquivo.is_file():
            z.write(arquivo,arcname=arquivo.name)
print('\nArquivos gerados sem nenhuma dependência de Drive pessoal:')
for p in sorted(SAIDA.iterdir()): print(' -',p.name)
print('\nZIP final:',zipfinal)
print('Não foram incluídos os três arquivos NetCDF brutos.')
```


## Considerações metodológicas

Os níveis de retorno e as probabilidades de excedência dos modelos não estacionários foram estimados utilizando os valores das covariáveis referentes a 2025.

As precipitações de 110,6 mm, 128,4 mm e 138,6 mm foram consideradas como cenários de magnitude aplicados aos pixels da Zona da Mata mineira.

Os períodos de retorno equivalentes representam as condições fixadas pelas covariáveis de referência e não constituem previsões do intervalo real de espera entre eventos futuros.

Para a reprodução independente, foi utilizada a série histórica pública ONI v5 da NOAA em substituição à planilha de El Niño empregada na análise original. A equivalência completa entre as duas séries ainda requer verificação.

A disponibilidade dos dados e a execução integral dependem do acesso aos repositórios públicos utilizados.
